## 1. ¿Qué es la inferencia bayesiana?

La estadística bayesiana es un enfoque de la inferencia estadística basado en el **Teorema de Bayes**. A diferencia de la estadística frecuentista, aquí la probabilidad representa un **grado de creencia** que se actualiza conforme se observan nuevos datos.

La idea central es:

$$P(\theta \mid X) = \frac{P(X \mid \theta)\, P(\theta)}{P(X)}$$

donde:
- $P(\theta)$ es la **distribución a priori** (lo que creemos antes de ver datos)
- $P(X \mid \theta)$ es la **verosimilitud** (qué tan probable es observar $X$ dado $\theta$)
- $P(X)$ es la **evidencia** (constante de normalización)
- $P(\theta \mid X)$ es la **distribución a posteriori** (nuestra creencia actualizada)


## 2. Ejemplo clásico: moneda sesgada

Queremos estimar la probabilidad $\theta$ de que una moneda caiga en cara.  
Lanzamos la moneda $n$ veces y observamos $k$ caras.

Usamos una distribución **Beta** como prior conjugado para la Binomial:

$$\theta \sim \text{Beta}(\alpha, \beta)$$

Después de observar $k$ caras en $n$ lanzamientos, el posterior es:

$$\theta \mid X \sim \text{Beta}(\alpha + k,\; \beta + n - k)$$


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Parámetros del prior (creencia inicial: moneda justa)
alpha_prior, beta_prior = 2, 2

# Datos observados
n = 20       # lanzamientos
k = 14       # caras

# Posterior
alpha_post = alpha_prior + k
beta_post  = beta_prior + (n - k)

# Visualización
theta = np.linspace(0, 1, 300)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(theta, beta.pdf(theta, alpha_prior, beta_prior),
        color='steelblue', linestyle='--', label=f'Prior  Beta({alpha_prior},{beta_prior})')
ax.plot(theta, beta.pdf(theta, alpha_post, beta_post),
        color='tomato', linewidth=2, label=f'Posterior  Beta({alpha_post},{beta_post})')
ax.axvline(k/n, color='gray', linestyle=':', alpha=0.7, label=f'MLE = {k/n:.2f}')
ax.set_xlabel('θ (P(cara))', fontsize=11)
ax.set_ylabel('Densidad', fontsize=11)
ax.set_title('Actualización Bayesiana — Estimación de probabilidad de cara', fontsize=12)
ax.legend()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()

print(f'Media a priori:    {alpha_prior/(alpha_prior+beta_prior):.3f}')
print(f'Media a posteriori: {alpha_post/(alpha_post+beta_post):.3f}')
print(f'MLE (frecuentista): {k/n:.3f}')
```

## 3. Propiedades del prior conjugado Beta-Binomial

| Distribución | Prior | Posterior |
|---|---|---|
| Binomial | Beta(α, β) | Beta(α+k, β+n−k) |
| Poisson | Gamma(α, β) | Gamma(α+k, β+n) |
| Normal (σ conocida) | Normal(μ₀, τ₀²) | Normal(μₙ, τₙ²) |

El prior conjugado tiene la ventaja de que el posterior es de la **misma familia** que el prior, lo que facilita el cálculo analítico.


## 4. Influencia del tamaño muestral

A medida que aumenta $n$, el posterior se concentra alrededor del valor verdadero independientemente del prior elegido — los datos "dominan" la creencia inicial.


```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=False)
escenarios = [(5, 3), (20, 14), (100, 67)]   # (n, k)

for ax, (n_i, k_i) in zip(axes, escenarios):
    a_post = alpha_prior + k_i
    b_post = beta_prior + (n_i - k_i)
    ax.fill_between(theta, beta.pdf(theta, alpha_prior, beta_prior),
                    alpha=0.25, color='steelblue', label='Prior')
    ax.fill_between(theta, beta.pdf(theta, a_post, b_post),
                    alpha=0.45, color='tomato', label='Posterior')
    ax.set_title(f'n={n_i}, k={k_i}', fontsize=11)
    ax.set_xlabel('θ')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    if ax == axes[0]:
        ax.set_ylabel('Densidad')
        ax.legend(fontsize=9)

plt.suptitle('Efecto del tamaño muestral sobre el posterior', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
```
