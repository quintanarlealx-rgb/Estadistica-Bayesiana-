En este notebook se replica de forma práctica el **Teorema del Límite Central (TLC)** mediante simulación, usando un experimento propio: el lanzamiento repetido de un dado.


## 1. Planteamiento del experimento

Sea $X$ el resultado de lanzar un dado justo de 6 caras.

$$X \sim \text{Uniforme Discreta}\{1,2,3,4,5,6\}$$

$$E[X] = 3.5, \qquad \text{Var}(X) = \frac{35}{12} \approx 2.917$$

Esta distribución es claramente **no normal** (es plana, discreta). El TLC predice que el promedio de $n$ lanzamientos, $\bar{X}_n$, se aproximará a una normal a medida que $n$ crece.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

rng = np.random.default_rng(7)

# Distribución original (un solo dado)
un_dado = rng.integers(1, 7, size=10000)

plt.figure(figsize=(6, 3.5))
plt.hist(un_dado, bins=np.arange(0.5, 7.5, 1), rwidth=0.8,
         color='steelblue', edgecolor='white')
plt.xticks(range(1, 7))
plt.title('Distribución de un solo dado (no normal)')
plt.xlabel('Resultado')
plt.ylabel('Frecuencia')
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

## 2. Simulación: promedio de $n$ dados

Simulamos miles de repeticiones del experimento "lanzar $n$ dados y calcular su promedio", para distintos valores de $n$.


```python
def promedio_n_dados(n, repeticiones=5000):
    """Simula 'repeticiones' veces el promedio de n dados."""
    tiros = rng.integers(1, 7, size=(repeticiones, n))
    return tiros.mean(axis=1)

valores_n = [1, 2, 5, 30]

fig, axes = plt.subplots(1, len(valores_n), figsize=(13, 3.5), sharey=False)

for ax, n in zip(axes, valores_n):
    promedios = promedio_n_dados(n)
    ax.hist(promedios, bins=30, density=True, color='steelblue',
            alpha=0.65, edgecolor='white', linewidth=0.4)

    # Curva normal teórica con la misma media y varianza
    mu, sigma = 3.5, np.sqrt(35/12 / n)
    x = np.linspace(promedios.min(), promedios.max(), 200)
    ax.plot(x, norm.pdf(x, mu, sigma), color='tomato', linewidth=1.8)

    ax.set_title(f'Promedio de {n} dado(s)', fontsize=10)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_yticks([])

plt.suptitle('Convergencia del promedio de dados a una distribución normal', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
```

## 3. Verificación numérica

Comparamos la media y desviación estándar empírica del promedio con lo que predice la teoría:

$$E[\bar{X}_n] = \mu = 3.5 \qquad \quad \text{SD}(\bar{X}_n) = \frac{\sigma}{\sqrt{n}}$$


```python
print(f'{"n":>5} | {"Media emp.":>12} | {"Media teo.":>12} | {"SD emp.":>10} | {"SD teo.":>10}')
print('-'*58)

sigma = np.sqrt(35/12)
for n in [1, 2, 5, 10, 30, 100]:
    promedios = promedio_n_dados(n, repeticiones=20000)
    sd_teo = sigma / np.sqrt(n)
    print(f'{n:>5} | {promedios.mean():>12.4f} | {3.5:>12.4f} | '
          f'{promedios.std():>10.4f} | {sd_teo:>10.4f}')
```

## 4. Conclusión

A pesar de que la distribución de un solo dado es **uniforme discreta** (totalmente distinta a una campana), el promedio de varios dados converge rápidamente a una distribución **normal**, confirmando empíricamente el TLC.

Esto es justamente lo que permite, en estadística bayesiana, usar aproximaciones normales para el posterior cuando se cuenta con un tamaño de muestra razonablemente grande.
